```python
# string pelindrom
def is_pelindrom(s):
  n=len(s)
  left=0
  right=n-1
  while left<right:
    if s[left] != s[right]:
      print("not pelindrome")
      return False
    left +=1
    right -=1
  print("pelindrome")
  return True
print(is_pelindrom("abcdecba"))
```

```python
def selectionSort(nums):
    for i in range(len(nums)):
        mini = i   # index of smallest

        for j in range(i+1, len(nums)):
            if nums[j] < nums[mini]:
                mini = j

        # swap
        nums[i], nums[mini] = nums[mini], nums[i]

    return nums


arr = [3,5,6,5,33,8,2,8,3]
print(selectionSort(arr))
```

```python
# insert at position


class Node:
  def __init__(self,val):
    self.val=val
    self.next=None
class SinglyLinkedList:
  def __init__(self):
    self.head=None

  # append
# tc=O(n) using self.tail pointer tc is O(1)
  def append(self,val):
    new_node=Node(val)
    if self.head==None:
      self.head=new_node
    else:
      curr=self.head
      while curr.next is not None:
        curr=curr.next
      curr.next=new_node

# insert
# tc=O(n)
# sc=O(1)
  def insert(self,val,pos):
    count=0
    new_node=Node(val)
    curr=self.head
    while count <pos-1:
      curr=curr.next
      count+=1
    new_node.next=curr.next
    curr.next=new_node

# traverse linked list
# tc=O(n)
# sc=O(1)
  def traverse(self):
    curr=self.head
    while curr is not None:
      print(curr.val,end="->")
      curr=curr.next
    print("None")

# object
sll = SinglyLinkedList()

# append first
arr = [2,6,7,4,7]

for i in arr:
    sll.append(i)

# then insert
sll.insert(3,2)

# display
sll.traverse()
```
